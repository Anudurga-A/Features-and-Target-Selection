```python
import pandas as pd
from sklearn.datasets import load_iris
```

```python
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["target"] = iris.target

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
X = df[
    [
        "sepal length (cm)",
        "sepal width (cm)",
        "petal length (cm)",
        "petal width (cm)"
    ]
]

y = df["target"]

print("Selected Features:")
print(X.columns.tolist())

print("\nSelected Target:")
print(y.name)
```

```python
print("FEATURE SELECTION EXPLANATION")

print("""
The four input features represent the measurements
of the Iris flower.

These features are selected because they help
identify and classify different Iris flower species.

The target column is 'target', which represents
the flower species encoded as numerical values.

The target is not included in the input features
because it is the value we want to predict.
""")
```

```python
print("RESULT:")
print("Features and target were selected successfully.")
print("The dataset is ready for supervised machine learning.")
```
